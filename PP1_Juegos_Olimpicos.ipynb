El objetivo es explorar características relacionadas con la participación y el desempeño de los países, buscando identificar patrones entre la cantidad de atletas participantes, las medallas obtenidas y otras condiciones que podrían estar asociadas con el rendimiento, como la temporada en que se desarrollan los Juegos y el hecho de competir como país anfitrión.

El conjunto de datos utilizado se encuentra en un archivo en formato **.CSV** y contiene información histórica sobre la participación de atletas en los Juegos Olímpicos. El dataset completo posee más de **315.000 registros y 15 variables**.

Finalmente, se estudiará el posible **efecto de la localía**, comparando la cantidad de medallas obtenidas por los países cuando fueron anfitriones de los Juegos con las obtenidas en otras ediciones que no fueron sede.

### Tratamiento de NOC

```python
# Filtrar los registros con valores nulos en NOC
noc_nulos = df[df["noc"].isnull()].copy()
noc_nulos.shape
```

```python
# Contar los NOC nulos por año
cantidad_por_anio = noc_nulos["anio"].value_counts()

# Ordenar los años de menor a mayor
cantidad_por_anio = cantidad_por_anio.sort_index()

cantidad_por_anio
```

```python
# Obtener los IDs de atletas que tienen un NOC conocido
ids_con_noc = df[df["noc"].notnull()]["id"].unique()

# Verificar si los atletas con NOC nulo aparecen en otros registros con NOC conocido
noc_nulos["id"].isin(ids_con_noc).value_counts()
```

```python
# Trabajar solamente con los registros que tienen NOC conocido
datos_con_noc = df[df["noc"].notnull()]

# Contar cuántos NOC diferentes tiene asociado cada atleta
noc_por_id = datos_con_noc.groupby("id")["noc"].nunique()

# Ver cuántos atletas tienen 1, 2, 3... NOC diferentes
noc_por_id.value_counts()
```

```python
# Obtener los IDs de atletas asociados a un único NOC
ids_noc_unico = noc_por_id[noc_por_id == 1].index

# Ver cuántos registros con NOC nulo pertenecen a atletas con un único NOC conocido
noc_nulos["id"].isin(ids_noc_unico).value_counts()
```

```python
# Identificar atletas que tienen un único NOC conocido
condicion_id = df["id"].isin(ids_noc_unico)

# Identificar registros donde el NOC no es nulo
condicion_noc = df["noc"].notnull()

# Filtrar las filas que cumplen ambas condiciones
noc_unico = df[condicion_id & condicion_noc]

# Conservar solamente las columnas necesarias
noc_unico = noc_unico[["id", "noc"]]

noc_unico.head()
```

```python
# Eliminar combinaciones repetidas de ID y NOC
noc_unico = noc_unico.drop_duplicates()
noc_unico.head(10)
```

```python
noc_unico.shape
```

```python
#Filtramos solo filas con medalla y eliminamos duplicados por evento,
# año/noc/medalla, con esto logramos que los deportes que se jueguen en
# equipo cuenten como una unica medalla para el pais correspondiente
df_medallas_unicas = df.dropna(subset=['medalla']).drop_duplicates(
    subset=['anio', 'temporada', 'noc', 'evento', 'medalla']
)

#Agrupamos medallas reales ganadas por pais
df_total_medallas = df_medallas_unicas.groupby('noc')['medalla'].count().reset_index(
    name='medallas_reales'
)

#Agrupamos total de atletas unicos por pais
df_total_atletas = df.groupby('noc')['id'].nunique().reset_index(
    name='total_atletas'
)

#Unimos ambos DataFrames para poder calcular la eficiencia
df_h2 = pd.merge(df_total_atletas,
                 df_total_medallas,
                 on='noc',
                 how='left').fillna(0)
df_h2['eficiencia'] = (df_h2['medallas_reales'] / df_h2['total_atletas']).round(4)

#Ordenamos por eficiencia de mayor a menor
df_h2 = df_h2.sort_values(by='eficiencia', ascending=False)

#Verificamos que lo que hicimos es correcto
df_h2.head(10)
```

Luego del análisis de los datos y de los gráficos realizados, podemos observar una diferencia importante en la cantidad promedio de medallas obtenidas por los países según su condición de localía. Los países analizados obtuvieron, en promedio, 36,89 medallas cuando fueron anfitriones, frente a 20,65 medallas en las ediciones en las que participaron sin serlo. Esto representa una diferencia de aproximadamente 16,24 medallas por edición.

Si bien existen casos particulares de países cuyos resultados no siguen la tendencia general, la comparación muestra que, en el conjunto analizado, la cantidad promedio de medallas es mayor cuando los países compiten como locales.

Por lo tanto, los resultados obtenidos respaldan la hipótesis planteada. Sin embargo, esta diferencia debe interpretarse como una asociación observada en los datos y no como una demostración de que la condición de anfitrión sea, por sí sola, la causa del aumento en la cantidad de medallas.
